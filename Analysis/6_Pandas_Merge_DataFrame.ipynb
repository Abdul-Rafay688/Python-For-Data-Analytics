```python
import pandas as pd
import matplotlib.pyplot as plt
import datasets as load_datasets

#Loading data
datasets = load_datasets.load_dataset('lukebarousse/data_jobs')
df = datasets['train'].to_pandas()

#Data cleaning
df['job_posted_date'] = pd.to_datetime(df['job_posted_date'])
```

```python
df_US = df[df['job_country'] == 'United States']
```

```python
df_US['job_posted_month'] = df_US['job_posted_date'].dt.strftime('%B')
```

```python
df_Pivot = df_US.pivot_table(
    index='job_posted_month',
    columns='job_title_short',
    aggfunc='size'
)
```

```python
df_Pivot.reset_index(inplace=True)

df_Pivot['month_no'] = pd.to_datetime(df_Pivot['job_posted_month'], format='%B').dt.month
df_Pivot.sort_values('month_no', inplace=True)
df_Pivot.set_index('job_posted_month', inplace=True)
df_Pivot.drop(columns='month_no', inplace=True)
```

```python
df_Pivot
```

```python
df_software = pd.read_csv("https://lukeb.co/software_csv", index_col='job_posted_month')

df_software
```

```python
df_US_merged = df_Pivot.merge(df_software, on='job_posted_month')
df_US_merged
```

```python
top_5 = (df_US_merged.sum().sort_values(ascending=False).head().index.to_list())
top_5
```

```python
df_US_merged[top_5].plot(kind='line')
plt.title('Top 5 Job Postings per Month in US')
plt.xlabel('2023')
plt.ylabel('Job Counts')
plt.ylim(0, 20000)
plt.legend()
plt.show()
```
