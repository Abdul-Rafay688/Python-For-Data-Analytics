```python
import pandas as pd
import matplotlib.pyplot as plt
import datasets as load_datasets

#Loading data
datasets = load_datasets.load_dataset('lukebarousse/data_jobs')
df = datasets['train'].to_pandas()

#Data cleaning
df['job_posted_date'] = pd.to_datetime(df['job_posted_date'])
```

```python
median_yearly_salary = df['salary_year_avg'].median()
median_hourly_salary = df['salary_hour_avg'].median()
```

```python
df_filled = df
df_filled['salary_year_avg'] = df_filled['salary_year_avg'].fillna(median_yearly_salary)
df_filled['salary_hour_avg'] = df_filled['salary_hour_avg'].fillna(median_hourly_salary)
```

```python
df.loc[0:10, 'salary_year_avg':'salary_hour_avg']
```

```python
#Fillteration With any Subset of Columns
df_unique = df_filled

df_unique = df_unique.drop_duplicates()

print("Original DataFrame shape: ",len(df_filled))
print("Unique DataFrame shape: " ,len(df_unique))
print("Number of duplicate rows removed: " ,len(df_filled) - len(df_unique))
```

```python
#Fillteration With any Subset of Columns
df_unique = df_filled

df_unique = df_unique.drop_duplicates(subset=['job_title', 'company_name'])

print("Original DataFrame shape: ",len(df_filled))
print("Unique DataFrame shape: " ,len(df_unique))
print("Number of duplicate rows removed: " ,len(df_filled) - len(df_unique))
```
